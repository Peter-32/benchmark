```python
is_valid_release = replay.release_string >= '5.0.16'
assert is_valid_release
```

```python
player1 = 'Serral'
player2 = None
player1_race = 'Zerg'
player2_race = None
for player in replay.players:
    if player.name == 'Serral':
        assert player.play_race == 'Zerg'
    else:
        player2 = player.name
        player2_race = player.play_race
print(player1, player2, player1_race, player2_race)
```

```python
for event in replay.events:
    if event.second < 120:
        continue

    if hasattr(event, "player") and event.player:
        if event.player.name == 'Serral':
            print(event.second, event.name, event)
    
    if event.second > 240:
        break
```
